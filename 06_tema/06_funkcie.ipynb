# Funkcie v Pythone

## 1. Čo je to funkcia?

Funkcia je **pomenovaný blok kódu**, ktorý môžeme opakovane volať.

### Syntax funkcie:

```python
def nazov_funkcie(parameter1, parameter2):
    # kód funkcie
    # odsadenie je dôležité!
    return vysledok
```

- `def` — kľúčové slovo, ktorým začína každá funkcia
- `nazov_funkcie` — meno, pod ktorým funkciu voláme
- `parameter1, parameter2` — vstupné dáta (môže byť 0, 1, 2 alebo viac)
- `return` — vráti výsledok; keď Python narazí na `return`, funkcia končí


## 2. Najjednoduchšia funkcia — nerobí nič

````python
# Zavolanie funkcie
nerob_nic()
````

## 4. Funkcia s jedným parametrom

Táto funkcia prijíma jeden parameter a vracia vypočítanú čistú sumu (predpokladáme daň 23 %).

Aby mohla správne fungovať, musíme jej poskytnúť parameter typu `int` alebo `float`.


````python
def netto(brutto):
    return round(brutto / 1.23, 2)
````

````python
# Volanie funkcie
netto(215.27)
````

````python
# Volanie s pomenovaným parametrom
netto(brutto=215.27)
````

### Úloha 3: Oprav chybu v kóde

## 5. Funkcia s dvoma parametrami

Táto funkcia prijíma dva parametre a vracia čistú sumu vypočítanú na ich základe.

Predpokladáme, že daň bude zadaná ako číslo typu `float` v rozsahu od 0 do 1.


````python
def netto(brutto):
    return round(brutto / 1 + 0.23, 2)
````

````python
def netto(brutto, dan):
    return round(brutto / (1 + dan), 2)
````

````python
# Rôzne spôsoby volania
print(netto(215.25))
print(netto(215.25, dan=0.23))
print(netto(brutto=215.25, dan=0.23))
````

## 6. Funkcie s predvolenými hodnotami parametrov

Druhý parameter môže byť vynechaný: v takom prípade sa použije jeho predvolená hodnota.

Predpokladáme, že daň (ak je zadaná) bude číslo typu `float` v rozsahu od 0 do 1.


````python
def netto(brutto, dan=0.23):
    return round(brutto / (1 + dan), 2)
````

````python
# Všetky tieto spôsoby fungujú
print(netto(215.25, 0.23))   # zadaná daň
print(netto(215.25))          # použije predvolenú 0.23
print(netto(100, 0.08))   # iná daň
````

### Úloha 6: Doplň predvolenú hodnotu

Uprav funkciu `pozdrav()` tak, aby mala predvolenú hodnotu pre meno `"svet"`.


````python
# Doplň predvolenú hodnotu parametra meno
def pozdrav(meno = 'svet'):
    return f"Ahoj, {meno}!"
````

````python
# Test
print(pozdrav("Peter"))  # Malo by vypísať: Ahoj, Peter!
print(pozdrav())          # Malo by vypísať: Ahoj, svet!
````

## 7. Návratová hodnota z funkcie

Keď Python narazí na kľúčové slovo `return`, funkcia končí a hodnota za slovom `return` sa stane výsledkom funkcie.

### Výsledok funkcie môžeme použiť rôznymi spôsobmi:


````python
# Výsledok sa stane hodnotou premennej
cista_cena = netto(215.25)
print(cista_cena)
````

````python
cista_cena
````

````python
# Výsledok sa použije vo výpočte
rocne_naklady = netto(215.25) * 365
print(rocne_naklady)
````

### Úloha 7: Oprav chybu s return

Nasledujúca funkcia má problém s umiestnením `return`. Oprav ho!


````python
# Táto funkcia má zle umiestnený return - oprav to!
def sucet_cisel(a, b):
    vysledok = a + b
    return vysledok

print(sucet_cisel(5, 3))  # Malo by vypísať 8
````

## 8. Volanie funkcií s premennými

Doteraz sme funkciám odovzdávali hodnoty zadané priamo pri volaní.
Parametre ale môžu pochádzať aj z premenných alebo byť vypočítané.


````python
cena1 = 100
cena2 = 250

print("Čistá cena 1:", netto(cena1))
print("Čistá cena 2:", netto(cena2))
print("Čistá cena spolu:", netto(cena1 + cena2))
````

### Úloha 8: Použi funkciu s premennými


Vytvor funkciu `prepocet_meny(suma, kurz=1.08)`, ktorá prepočíta sumu podľa zadaného kurzu.
Potom ju použi na prepočet troch súm: 50, 120 a 300.


````python
def prepocet_meny(suma, kurz = 1.08):
    return round(suma/kurz, 2)
````

````python
s1 = 50
s2 = 120
s3 = 300
# Zavolaj funkciu pre každú sumu
````

````python
prepocet_meny(s1)
````

````python
prepocet_meny(s2)
````

````python
prepocet_meny(s3)
````
