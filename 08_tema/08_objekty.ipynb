# Úvod do objektovo orientovaného programovania v Pythone

V tejto lekcii sa naučíte základy **objektovo orientovaného programovania** (OOP) v Pythone.

## 1. Čo je objektovo orientované programovanie?

**Objektovo orientované programovanie** je spôsob, ako písať programy pomocou **objektov**.

- **Objekt** spája **dáta** (informácie) a **správanie** (čo vie robiť)
- Program sa skladá z objektov, ktoré spolu komunikujú

### Príklad zo života:
- **Auto** je objekt
- Dáta: farba, značka, rýchlosť, množstvo paliva
- Správanie: môže jazdiť, brzdiť, zatáčať, tankovať

## 2. Trieda vs. Objekt

V Pythone rozlišujeme dva dôležité pojmy:

### **Trieda (Class)**
- Je to **predpis** alebo **šablóna**
- Definuje, ako bude objekt vyzerať a čo bude vedieť
- Je ako **výkres** alebo **recept**

### **Objekt (Instance)**
- Je to **konkrétny exemplár** triedy
- Každý objekt má svoje vlastné dáta
- Je ako **konkrétny výrobok** podľa výkresu

### Analógia:
- **Trieda** = Výkres domu
- **Objekty** = Konkrétne domy postavené podľa tohto výkresu

### Príklad, ktorý už poznáte!

- `list` je **trieda**
- `zoznam1` a `zoznam2` sú objekty (**inštancie**) triedy `list`

````python
zoznam1 = [1, 2, 3]
zoznam2 = ['a', 'b', 'c']

# Každý objekt má svoje vlastné dáta
print(zoznam1)
print(zoznam2)
````

````python
# Ale oba vedia rovnaké veci (metódy)
zoznam1.append(4)
zoznam2.append('d')
````

````python
print(zoznam1)
print(zoznam2)
````

````python
tup = (1, 2)
````

````python
tup.count(2)
````

## 3. Čo sú metódy?

**Metódy** sú funkcie, ktoré patria k objektu. Definujú, čo objekt **vie robiť**.

### Syntax volania metódy:
```python
objekt.metoda(argumenty)
```

### Príklady metód, ktoré už poznáte:

````python
# Trieda list má veľa metód
cisla = [3, 1, 4, 1, 5]

# Metóda sort() - zoradí zoznam
cisla.sort()
print('Zoradený zoznam:', cisla)
````

````python
# Metóda reverse() - obráti poradie
cisla.reverse()
print('Obrátený zoznam:', cisla)
print(cisla.reverse())
````

````python
# Metóda count() - spočíta výskyty
pocet = cisla.count(1)
print('Počet jedničiek:', pocet)
````

napr. `sort()` je metóda pre objekt, nie je všeobecná v base knižnici... prečo?

````python
# Trieda tuple
t = (3, 2, 5, 12, 2, 5, 5)
````

````python
t.index(5)
````

````python
t.count(5)
````

````python
t.sort()
````

## 4. Metódy s rôznymi triedami

Rôzne triedy majú rôzne metódy, pretože vedia robiť rôzne veci.

### Trieda `str` (reťazce)

Objekty typu `str` vedia pracovať s textom:

````python
text = 'python programovanie'

# Metóda upper() - prevedie na veľké písmená
text.upper()
````

````python
# Metóda title() - začne každé slovo veľkým písmenom
text.title()
````

````python
# Metóda replace() - nahradí časť textu
text.replace('python', 'Python')
````

````python
# Metóda split() - rozdelí text na zoznam
text.split()
````

### Trieda `dict` (slovníky)

Objekty typu `dict` vedia pracovať s pármi kľúč-hodnota:

````python
student = {
    'meno': 'Jan',
    'vek': 20,
    'odbor': 'Informatika'
}
````

````python
# Metóda keys() - vráti všetky kľúče
print('Kľúče:', student.keys())

# Metóda values() - vráti všetky hodnoty
print('Hodnoty:', student.values())

# Metóda items() - vráti páry kľúč-hodnota
print('Položky:', student.items())

# Metóda get() - získa hodnotu (bezpečne)
print('Meno:', student.get('meno'))
print('Email:', student.get('email', 'Nie je uvedený'))
````

## 5. Prečo existuje viac tried?

Každá trieda je optimalizovaná pre iný účel:

| Trieda | Účel | Príklad použitia |
|-------|------|----------------|
| `list` | Zoradená kolekcia prvkov | Zoznam známok, nákupný zoznam |
| `dict` | Ukladanie dát s popiskami (kľúčmi) | Telefónny zoznam, nastavenie programu |
| `str` | Práca s textom | Mená, vety, texty |
| `set` | Kolekcia unikátnych prvkov | Zoznam navštívených miest |
| `tuple` | Nemenná zoradená kolekcia | Súradnice, konštantné dáta |

## 6. Zistenie typu objektu a dostupných metód

Python nám dáva nástroje, ako zistiť viac o objektoch:

### Metódy pre list

````python
cisla = [1, 2, 3]
````

````python
# Funkcia type() - zistí typ objektu (triedu)
print('Typ cisla:', type(cisla))
````

````python
cisla.
````

Funkcia `dir`

````python
len(cisla)
````

````python
cisla.
````

````python
[metoda for metoda in dir(cisla) if not metoda.startswith('_')]
````

````python
# Funkcia dir() - zobrazí všetky metódy objektu
print('\nMetódy list:')
print([m for m in dir(cisla) if not m.startswith('_')])
````

### Metódy pre reťazec

````python
text = 'Python'
print('Typ text:', type(text))
````

````python
# Funkcia dir() - zobrazí všetky metódy objektu
print('\nMetódy pre reťazec:')
print([m for m in dir(text) if not m.startswith('_')])
````
